```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
data = {
    "Colour": ["Red", "Green", "Red", "Green", "Red", "Green"],
    "Size": [8, 7, 9, 6, 8, 7],
    "Texture": ["Soft", "Hard", "Soft", "Hard", "Soft", "Hard"],
    "Class": ["Ripe", "Unripe", "Ripe", "Unripe", "Ripe", "Unripe"]
}
df = pd.DataFrame(data)

df
```

```python
print("Dataset shape:", df.shape)
print("\nDataset information:")
print(df.info())

print("\nClass distribution:")
print(df["Class"].value_counts())
```

```python
df["Class"].value_counts().plot(kind="bar")

plt.title("Tomato Class Distribution")
plt.xlabel("Tomato Class")
plt.ylabel("Number of Tomatoes")
plt.show()
```
